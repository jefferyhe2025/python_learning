## 综合练习：商店销量与单价

数据在 `data_base/data/ch1/`：

### 题目 1：按季度汇总销量

一年 12 个月 → 4 个季度，每季度 3 个月。

思路：把 `(20, 12)` 看成 `(20, 4, 3)`，再对最后一维（季度内的 3 个月）求和，得到 `(20, 4)`。

```python
import numpy as np
from pathlib import Path

BASE = Path("data_base/data/ch1")

# 把 200 个文件叠成一个大数组：第 0 维=商品，第 1 维=年，第 2 维=月
sales = np.stack(
    [np.load(BASE / f"shop_sales/product_{i}.npy") for i in range(1, 201)]
)
print("销量数组 shape:", sales.shape)  # (200, 20, 12)

# 随便拿第 1 种商品做题目 1 的演示（题目 1 本质是「对某一个 (20,12) 数组」做季度汇总）
arr = sales[0]
arr.shape
```

```python
# (年, 月) -> (年, 季度, 季度内月份) -> 对季度内求和
quarter_sales = arr.reshape(20, 4, 3).sum(axis=-1)
print(quarter_sales.shape)  # (20, 4)
quarter_sales[:3]  # 看前 3 年
```

### 题目 3：还原 2001-01 ~ 2020-12 的全部单价

已知：
- 2015 年 1 月的单价（锚点）
- 每个月相对「上个月」的涨幅 `rate`

关系很简单：

$$\text{本月单价} = \text{上月单价} \times (1 + \text{本月涨幅})$$

从 2015-01 往前除、往后乘，就能推出全部 240 个月。最后整理成 `(年, 月, 商品) = (20, 12, 200)`。

```python
rate = np.load(BASE / "increase_rate.npy")       # (200, 240)
price_2015_01 = np.load(BASE / "unit_price.npy")  # (200,)

# 2001-01 是第 0 个月，2015-01 是第 (2015-2001)*12 = 168 个月
idx = (2015 - 2001) * 12  # 168

# 用 cumprod 一次性还原（比 for 循环更快、更干净）
# 约定：price[t] = price[t-1] * (1 + rate[t])，因此 rate[:, 0] 实际用不到
# cum[:, t] = (1+rate[1]) * ... * (1+rate[t])，且 cum[:, 0] = 1
ones = np.ones((200, 1))
cum = np.cumprod(np.concatenate([ones, 1 + rate[:, 1:]], axis=1), axis=1)

# 以 2015-01 为锚：任意月单价 = 锚点单价 * (该月累计因子 / 锚点累计因子)
prices_flat = price_2015_01[:, None] * cum / cum[:, [idx]]  # (200, 240)

# 变成题目要求的 (年, 月, 商品)
unit_prices = prices_flat.reshape(200, 20, 12).transpose(1, 2, 0)
print(unit_prices.shape)  # (20, 12, 200)

# 校验：2015 年是第 15 年（下标 14），1 月下标 0，应等于原始单价
np.allclose(unit_prices[14, 0], price_2015_01)
```

### 题目 4：每种商品「月销售额」最高的月份

月销售额 = 单价 × 销量。对每种商品在 240 个月里找最大值所在的月份，输出形如 `"2008-05"` 的字符串。

```python
# 对齐维度：sales (200,20,12)，unit_prices (20,12,200) -> 都变成 (200,20,12)
revenue = sales * unit_prices.transpose(2, 0, 1)  # (200, 20, 12)

# 把年月摊平成 240 列，找每行最大值的位置
flat = revenue.reshape(200, -1)          # (200, 240)
max_pos = flat.argmax(axis=1)            # 每种商品一个下标 0~239

years = 2001 + max_pos // 12
months = 1 + max_pos % 12
peak_months = np.array([f"{y:04d}-{m:02d}" for y, m in zip(years, months)])

print(peak_months.shape)  # (200,)
peak_months[:10]  # 前 10 种商品的峰值月份
```
