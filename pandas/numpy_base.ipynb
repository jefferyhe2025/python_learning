### 题目 3：还原 2001-01 ~ 2020-12 的全部单价

核心关系只有一句：

$$\text{本月单价} = \text{上月单价} \times (1 + \text{本月涨幅})$$

已知 2015-01 的单价当「锚点」：往后连乘、往前连除，得到全部 240 个月，再整理成 `(20, 12, 200)`。

两种常见写法对比：

| | 全局累计比（我之前的） | 分段往前/往后（参考写法） |
|--|--|--|
| 思路 | 先算整段 `cumprod`，再除以锚点处的累计值 | 锚点右边单独往后乘，左边单独往前除 |
| 优点 | 公式统一、代码短 | 方向一眼能看懂，只算需要的区间 |
| 缺点 | 要先拼一列 `1`，语义稍绕 | 往前那段有 `::-1` 翻转，初看容易懵 |

下面是整合优化版：保留「分段」的清晰语义，变量名说人话。

```python
rate = np.load("data_base/data/ch1/increase_rate.npy")  # (200, 240) 每月相对上月涨幅
base = np.load("data_base/data/ch1/unit_price.npy")     # (200,)     2015-01 单价（锚点）

# 月份下标：0=2001-01，168=2015-01，239=2020-12
n = (2015 - 2001) * 12  # 等价于 14 * 12
mult = 1.0 + rate         # 乘子：本月价 = 上月价 * mult[本月]

price = np.empty((200, 20 * 12))  # (商品, 240个月)
price[:, n] = base                # 先钉住锚点

# 往后（2015-02 ~ 2020-12）：从锚点出发连乘
# cumprod 后第 k 列 = mult[n+1] * mult[n+2] * ... * mult[n+1+k]
price[:, n + 1 :] = mult[:, n + 1 :].cumprod(axis=1) * base[:, None]

# 往前（2001-01 ~ 2014-12）：从锚点往回除 = 乘 1/mult
# 先把区间翻转，从近到远 cumprod，再翻转回来，对齐到月份 0..n-1
# rate[:, 0] 用不到：没有「更早一个月」可参照
inv = (1.0 / mult[:, 1 : n + 1])[:, ::-1].cumprod(axis=1)[:, ::-1]
price[:, :n] = inv * base[:, None]

# (商品,年,月) -> 题目要求的 (年,月,商品)
unit_prices = price.reshape(200, 20, 12).transpose(1, 2, 0)
print(unit_prices.shape)  # (20, 12, 200)
np.allclose(unit_prices[14, 0], base)  # 校验锚点月份
```

### 题目 4：每种商品「月销售额」最高的月份

月销售额 = 单价 × 销量。对每种商品在 240 个月里找最大值所在月份，输出形如 `"2008-05"`。

两种写法对比（结果相同）：

| | 我之前的 | 参考写法 |
|--|--|--|
| 对齐维度 | 把 `price` 转成 `(200,20,12)` 去乘销量 | 把销量转成 `(20,12,200)` 去乘 `price` |
| 摊平找最大 | `reshape(200, 240).argmax(1)`：一行一种商品 | `reshape(240, 200).argmax(0)`：一列一种商品 |
| 优点 | 「商品在第 0 维」和前面 `sales` 一致，好跟销量对读 | 一行写完更短；和第三题 `price` 形状直接相乘 |
| 缺点 | 稍多两行 | `transpose` + `argmax(0)` 要多想一步维度 |

整合版：沿用第三题的 `price` 形状直接相乘（更短），变量名说清楚，日期用 f-string。

```python
# sales/arr: (200,20,12)  单价 unit_prices/price: (20,12,200)
# 先把销量转成和单价一样的形状，相乘 = 每月销售额
revenue = sales.transpose(1, 2, 0) * unit_prices   # (20, 12, 200)

# 把「年×月」摊成 240 行，每种商品一列；沿时间轴找最大值下标
idx = revenue.reshape(-1, 200).argmax(axis=0)      # (200,) 取值 0~239

year = idx // 12 + 2001   # 第几个 12 月块 → 年份
month = idx % 12 + 1      # 块内余数 → 1~12 月

peak_months = [f"{y}-{m:02d}" for y, m in zip(year, month)]
peak_months[:5]  # 展示前 5 个
```
