```python
df = pd.DataFrame(
    data={
        'col_0':[1,2,3],
        'col_1':list('abc'),
        'col_2':[1.2,2.2,3.2]
    },
    index=[f'row_{i}' for i in range(3)]
)
df
```

```python
# 当索引一致时，序列的值直接对应填入DataFrame。
index = [f'row_{i}' for i in range(3)]
df = pd.DataFrame(
    data={
        'col_0':pd.Series([1,2,3],index=index),
        'col_1':list('abc'),
        'col_2':[1.2,2.2,3.2]
    },
    index=index
)
df
```

## 2.4窗口

#### 1.滑动窗口

##### 练一练：在Numpy中也有一个同名函数np.diff()，它与pandas中的diff功能相同吗？请查阅文档说明。

```python
#两者都在算「差分」，但：
# 默认相邻差：结果数值接近，pandas 会多出开头的 NaN
# n=2 / periods=2：含义不同，不能当成同一个函数用
# 记法：NumPy 的 n 是「做几遍」；pandas 的 periods 是「隔几格」。
```

```python
s = pd.Series([1,2,3,4,5])
s.rolling(3).apply(lambda x:list(x)[0])
```

```python
# 类滑窗函数shift()
s.shift(2)
```

```python
s.rolling(4).apply(lambda x:list(x)[-1]-list(x)[0])
```

```python
# 类滑窗函数diff()
s.diff(3)
```

```python
def my_pct(x):
    L = list(x)
    return L[-1] / L[0] - 1
s.rolling(2).apply(my_pct)
```

```python
s.pct_change()
```

##### 练一练：rolling对象的默认窗口方向都是向下滑动的，某些情况下用户需要逆向滑动的窗口，例如对[1,2,3]设定窗口为2的逆向sum操作，结果为[3,5,NaN]，此时应该如何实现？

```python
s = pd.Series([1,2,3])
s[::-1].rolling(2).sum()[::-1] # 翻转两次
```
