#### 2.DataFrame

```python
df = pd.DataFrame(
    data={
        'col_0':[1,2,3],
        'col_1':list('abc'),
        'col_2':[1.2,2.2,3.2]
    },
    index=[f'row_{i}' for i in range(3)]
)
df
```

##### 练一练：在上面的df中，如果data字典中'col_0'键对应的不是列表，而是1个索引与df中索引相同的Series，此时会发生什么？如果它的索引和df的索引不一致，又会发生什么？

```python
# 当索引一致时，序列的值直接对应填入DataFrame。
index = [f'row_{i}' for i in range(3)]
df = pd.DataFrame(
    data={
        'col_0':pd.Series([1,2,3],index=index),
        'col_1':list('abc'),
        'col_2':[1.2,2.2,3.2]
    },
    index=index
)
df
```

```python
# 当索引不一致时且Series中索引值唯一时，当前DataFrame行索引如果在Series中出现，
# 则用Series对应元素填充，否则设为缺失值。
df = pd.DataFrame(
    data={
        'col_0':pd.Series([1,2,3],index=['row_3','row_2','row_1']), # Nan、3、2
        'col_1':list('abc'),
        'col_2':[1.2,2.2,3.2]
    },
    index=index
)
df
```

```python
# 用[col_name]和[col_list]取出相应的列与由多个列组成的新的DataFrame
df['col_0']
```

```python
# New DataFrame
df[['col_0','col_1']]
```

##### 练一练：df['col_0']和df[['col_0']]二者得到的结果类型有什么区别？

```python
# 前者是Series，后者是DataFrame
df[['col_0']]
```

```python
# 把序列转换为列数为1的DataFrame
df['col_0'].to_frame()
```

```python
# 对DataFrame的行列进行转置
df.T
```

##### 练一练：给定一个DataFrame，请构造其转置且不得使用“.T”。

```python
df = pd.DataFrame(data={'A':[1,2,3],'B':[4,5,6]},index=list('abc'))
df_T = pd.DataFrame(df.values.T, index=df.columns, columns=df.index)
df.T.equals(df_T)
```

##### 修改或删除列

```python
df = pd.DataFrame(
    data={
        'col_0':pd.Series([1,2,3],index=index), # Nan、3、2
        'col_1':list('abc'),
        'col_2':[1.2,2.2,3.2]
    },
    index=index
)
df["col_0"] = df['col_0'].values[::-1] #颠倒顺序
df["col_2"] *= 2
df["col_3"] = ["apple","banana","cat"] #新增一列
df
```

```python
# 删除列
df.drop(["col_3"],axis=1)
```

```python
# 删除行
df.drop(["row_1"],axis=0)
```
