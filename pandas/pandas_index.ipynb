#### 5.query()函数

##### 练一练：
利用query()选出A学校和B学校的大一女生。

```python
df.query('(School in ["A","B"]) and '
         '(Gender=="Female") and '
         '(Grade=="Freshman")').head()
```

```python
# 列表的比较(==/!=) -> is in/not in
df.query('Grade == ["Junior","Senior"]').head()
```

```python
# "@" 引用外部变量
query_list = ["Junior","Senior"]
df.query('Grade == @query_list').head()
```

#### 4.索引运算

```python
df_set_1 = pd.DataFrame([[0,1],[2,3],[4,5]],
                        index=pd.Index(['a','b','a'],name='id_1'))
```

```python
df_set_2 = pd.DataFrame([[4,5],[2,6],[7,1]],
                        index=pd.Index(['b','b','c'],name='id_2'))
df_set_2
```

```python
df_set_in_col_1 = df_set_1.reset_index()
df_set_in_col_1
```

```python
df_set_in_col_2 = df_set_2.reset_index()
df_set_in_col_2
```

```python
# 两张表要比的那一列还没设成索引时，用 isin 也能做“交集筛选”
df_set_in_col_1[df_set_in_col_1.id_1.isin(df_set_in_col_2.id_2)]
```
