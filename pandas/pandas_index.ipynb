## 3.3常用索引方法

#### 1.索引层的交换和删除

```python
import pandas as pd
import numpy as np

np.random.seed(0)
L1,L2,L3 = ['A','B'],['a','b'],['alpha','beta']
my_index1 = pd.MultiIndex.from_product([L1,L2,L3],
                                       names = ('Upper','Lower','Extra'))
L4,L5,L6 = ['C','D'],['c','d'],['cat','dog']
my_index2 = pd.MultiIndex.from_product([L4,L5,L6],
                                       names = ('Big','Small','Other'))
df_ex = pd.DataFrame(np.random.randint(-9,10,(8,8),),
                     index = my_index1,
                     columns= my_index2)
df_ex
```

##### swaplevel方法（只能交换两层）

```python
df_ex.swaplevel(0,2,axis = 1).head() #列索引交换
```

##### reorder_levels方法

```python
df_ex.reorder_levels([2,0,1],axis=0).head() # 行索引交换
```

##### 删除索引

```python
df_ex.droplevel(1,axis=1).head()
```

```python
df_ex.droplevel([0,1],axis=0).head()
```

#### 2.索引属性的修改

```python
# rename_axis()
df_ex.rename_axis(index={'Upper':'Changed_row'})
```

```python
# rename()
df_ex.rename(columns={'cat':'not cat'},level=2).head()
```

```python
# 参数值为函数
df_ex.rename(index=lambda x:str.upper(x),level=2).head()
```

```python
# 对索引整个元素的替换
new_values = iter(list('abcdefgh'))
df_ex.rename(index=lambda x:next(new_values),level=2).head()
```

```python
# map方法
df_temp = df_ex.copy()
df_temp.index = df_temp.index.map(lambda x:(x[0],x[1],str.upper(x[2])))
df_temp.head()
```

```python
# 多级索引的压缩
df_temp = df_ex.copy()
df_temp.index = df_temp.index.map(lambda x:(x[0]+'-'+x[1]+'-'+x[2]))
df_temp.head()
```

```python
# 展开
new_idx = df_temp.index.map(lambda x:tuple(x.split('-')))
df_temp.index = new_idx
df_temp.head()
```

##### 练一练：将上述df_temp的内层列索引合并到与外层列索引（用下划线连接）后删除内层行索引，即结果表压缩为两层列索引，同时再把外层行索引的索引名重命名为“Big_Other”。

```python
df_temp.columns = df_temp.columns.map(lambda x:(x[0]+'_'+x[2],x[1],x[2]))
df_temp.droplevel(-1,axis=1).rename_axis(columns = {'Big':'Big_Other'}).head()
```
